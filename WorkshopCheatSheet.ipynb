The syntax for assigning and reassigning the value is of the form `variable_name = value`. Some examples of variable assignment are presented below!

````python
# Setting up Variables
some_number = 11                # First time, just need to do
                                # name you want = thing you want it to be!

some_number = 31                # Can later change the value the same way!

some_number = some_number + 1   # Can use current value to set next value - this
                                # sets some_number to be some_number + 1.

some_number = "canary birds"    # Can have the variable hold anything - i.e,
                                # doesn't always have to have numbers if it
                                # first held a number, for example.
````

Let's address a common source of confusion in coding really quickly. Consider the below bit of code. Can you tell me what the value of `num_1` is going to
be without running it?

````python
num_1 = 5
num_2 = 10
num_1 = num_2
num_2 = num_2 + 5

print(f"The value of num_1 is {num_1}, and num_2 is {num_2}")
````

If you guessed that `num_1` was going to contain 15, don't worry, this is a common error! In programming, a statement like `num_1 = num_2` doesn't permanently mean that `num_1` and `num_2` will share the same value, regardless of what the value of `num_2` is. Rather, it just means that at that line, we set the value of `num_1` to be whatever the value of `num_2` is. Further modifications of `num_2` will not affect whatever was in `num_1`!!

Certain operations can have multiple different meanings, depending on the data type. For example, the + symbol performs regular addition on numbers of any kind, whereas for strings it joins them together. For example, run the below code and observe the output.

````python
# Sometimes Different Types Use The Same Operator in Different Ways

# + with ints or floats adds
# them together like we expect
variable_1 = 5
variable_2 = 3.1

print("Decimal '+' Result: ")
print(variable_1 + variable_2)
print()

# + with text joins it together
# (formally: concatenates it)!
variable_1 = "hello"
variable_2 = "there"

print("String '+' Result: ")
print(variable_1 + variable_2)
````

## If Statements

Currently, we don't have a way of taking multiple paths through our code, choosing to do a certain action or not, or anything similar. For example, what if I wanted to write something that took in a number, and only printed it back out if it was even? The way we do this is with what is known as "if statements", and a basic one is presented below. Try it out!

````python
# If Statement Demonstration
# Program prints whether input num is odd or even.

num = int(input("Please enter an integer: "))

if num % 2 == 0:
  print("Even")
else:
  print("Odd")

print("All done!")
````

How exactly does this work? Let's break this down line by line.

First off, the line `if num % 2 == 0:`. `num % 2` is calculated by the computer, and then is compared against the value of 0 to see if they are equal or not. Python uses `==` for comparing whether two values are equal, since using one equal sign is already reserved for assigning.

If the condition is true, we go into the first branch of the if statement, print "Even", then leave (there's nothing else to do in this branch) and go to the next shared line of code - printing "All done". If the condition fails, the code goes into the `else` branch, prints "Odd", leaves as there's nothing else to do, and once again goes to the next shared code line.



How can we tell that `print("Even")` belongs to the `if` branch, and `print("Odd")` belongs to the `else` branch? Well, Python uses the indentation of code lines to determine what belongs where.

`print("Even")` is indented in by a tab, so it belongs to the `if` branch, and only runs if we go into that branch. `else` is not indented, so doesn't belong to the `if` branch - as there's no more indented code in this section, that's the end of the `if` branch.

Likewise, `print("Odd")` is indented in by a tab, so it belongs to the `else` branch, and only runs if we go in it. As `print("All done!")` is not indented, it belongs to neither branch, so gets run always.

We can check any number of options at once in an if statement! If we want to check for > 2 branching paths, we need to use an `elif` (a contraction of "else if") statement and add an extra condition to check. We can have any number of `elif` statements. Also to note, `else` statements are completely optional; if there's no action you take if a condition isn't true, there's no need to write one. You can also have an army of `elif`s and no `else`, that is also perfectly valid!

You don't just need to check equality, too: you can check if something is greater than/less than, greater than equal to/less than equal to, and not equal as well. Some examples of all of this are presented below.

````python
# Elif + Other Testing Option Example

num = float(input("Please enter a number: "))

if num >= 10:                               # less than or equal to would be <=
  print("Bigger than or equal to 10")
elif num < 5:
  print("Smaller than 5")
elif num != 6.66:                           # if num is not equal to 6.66
  print("Somewhat small, but not demonic.")
else:
  print("😈")
````

We can also test 2 or more things in one condition by using the keywords `and`, `or` and `not`, to specify exactly what conditions we want. One thing to note: if you use several together, it is best to use parentheses to indicate exactly what logic you want. It's very easy to make mistakes otherwise, or interpret a logical condition differently to how Python evaluates it while the program is running.

“I want a bouquet with (blue cornflowers and black roses) or (yellow roses)”
is different to
“I want a bouquet with blue cornflowers and (black roses or yellow roses)”!!

````python
# Example of How to Use "or"! "and" and "not" are Used Very Similarly.
num = float(input("Please enter a number: "))

if num > 10 or num < 0:
  print("Needs more than one digit to print!")
````

## Functions

Often when we code, there's going to be a specific task that we need to repeat multiple times - for example, performing the same type of calculation over and over again. We could just copy and paste the same lines of code; however, this can get messy + hard to keep track of. You can forget to modify one out of 20 of those sections when updating your code, for example, and now everything's on fire and you don't know why.

A better option is to create a function - basically, a callable chunk of code you give a name to and can run whenever you want. `print()` and `input()` are some functions which we have already seen, ones that have been provided to us. An example of a user-written function is presented below.

````python
# Example of Function Implementation and Function Calling

# NOTE: these don't have to be in the same code block in Colab, BUT you do
# need to run the block containing the function before you run a block calling
# it!!



# Function Implementation

# keyword "def" used to indicate that it's a function
# Need to indent to show what lines belong to the functions and what don't!

def find_factorial(number):   # Inside the brackets you choose + specify
  index = 1                   # "parameters" - variables you pass in while
  curr_fact = 1               # calling!

  while index <= number:
    # Calculates curr_fact as 1 * 2 * ... * number
    curr_fact = curr_fact * index
    index = index + 1

  return curr_fact            # "return" keyword is necessary to specify what
                              # the function spits back out!




# Function Calling
num = int(input("Enter a number: "))

# Here, num is called an argument - it's what's being passed in to fill a
# parameter!
factorial = find_factorial(num)
print(f"The factorial of {num} is {factorial}.")      # we can also pass in raw
print(f"The factorial of 4 is {find_factorial(4)}.")  # data, not just variables!
````

The code inside a function doesn't run unless you specifically call it. Like with everything else we've seen so far in Python, we need to indent to show what lines of code belong to a function, and which lie outside of it. When writing a function you can give it any name that isn't currently in use by another function or variable (i.e, don't write another function named `print()`).

(Something to note - parameters are the variables we specify when we define the function, but arguments are what we pass in. This can be easy to confuse! In the above, `number` is a parameter, whereas `num` and `4` are arguments because they're the things being passed into the function. ### Functions With Multiple Parameters


When we create a function, we can give it any number of parameters (variables we pass in when we call the function). However, if we have more than one variable, there are multiple ways of going about actually passing in the values we want the parameters to have when we call.

The easiest is by position: for example, calling a function with a header of
```python
def raise_to_power(number, exponent):
```
as
```python
result = raise_to_power(5, 2)
```
sets the value of number to 5 and the value of exponent to 2. (Here, `number` and `exponent` are function parameters, and `5` and `2` are arguments to the function!)

We can also pass in the arguments by the names of the parameters; the next line is equivalent to the one we just wrote above!
```python
result = raise_to_power(number=5, exponent=2)
```

When we pass in arguments by parameter name, the order of them doesn't actually matter. If there are any arguments you are passing in by order, make sure to do that before any arguments you pass in by name! The below would return the same as the prior two function calls.
```python
result = raise_to_power(exponent=2, number=5)
```

Function parameters also can have default values. If I change the function header to the following,
```python
def raise_to_power(number, exponent=1):
```
then if I only pass in an argument for number, exponent assumes a default number of 1 and the function can still run as normal. However, if I was to change the function header to the following
```python
def raise_to_power(number=2, exponent):
```
and only pass in one argument positionally, then it would change the value of number, not exponent, and the function would crash. Make sure to put any parameters that have default values after parameters that do not when writing the function definition!

Working examples of all of the above can be seen below. Run them and see!

````python
# Multiple Parameter Demonstration

# You can just do this using **, but this is to demonstrate a function
# with multiple parameters!
def raise_to_power(number, exponent=1):
  index = 1
  curr_raised_num = number

  while index < exponent:
    curr_raised_num = curr_raised_num * number
    index = index + 1

  return curr_raised_num


# Raises 5 to the 2nd power - passing arguments in using parameter order! (Prints 25)
print(f"raise_to_power(5, 2) = {raise_to_power(5, 2)}")

# Order doesn't matter if we're passing in using the parameter name! (Prints 25)
print(f"raise_to_power(number=5, exponent=2) = {raise_to_power(number=5, exponent=2)}")
print(f"raise_to_power(exponent=2, number=5) = {raise_to_power(exponent=2, number=5)}")

# This will just print 5 since the default value for exponent is 1
print(f"raise_to_power(5) = {raise_to_power(5)}")
````

### Functions and Variable Scope

Any variables that we define inside a function cannot be accessed after we return from a function - whatever data got put in them last time we called the function gets scrubbed. If you need to access multiple things after a function's return, you need to return multiple things! You can do so simply using commas like in the example below:

````python
# Multiple Variable Return Demonstration

# A randomiser function that returns two
# variables at once!
def randomiser_function(num_1, num_2):
  output_1 = num_1 * num_2 + num_1/num_2
  output_2 = num_2 ** num_1 * (num_1 - num_2)

  return output_1, output_2

# output_1 goes into random_1, output_2
# goes into random_2
# Order MATTERS here - must match up!
random_1, random_2 = randomiser_function(4, 2)
print(f"random_1 = {random_1}")
print(f"random_2 = {random_2}")
````

You *can*, however, access variables that are defined outside of a function inside one. However, **PLEASE DON'T**. Doing this is a very easy way to write code that accidentally depends on some external variable you forgot about, that may work fine one time you run the notebook but fall apart once you reload it. If you need to access a variable from outside of a function, just add an extra parameter to the function + pass it in.

Some examples are shown below, and how we can access a specific element of a list.

Lists have quite a few helpful inbuilt methods if you need to work with them! To note - unless specified, **THEY MODIFY THE ORIGINAL LIST ITSELF**, and they don't just return a copy!

Here's a selection of useful methods with code examples of them!

|Method Name + Syntax|Description|
|-|-|
|`.append(<thing to insert>)`|Appends the given thing to the end of the list|
|`.index(<item in list>)`|Returns the index of the first matching item in the list, and chucks an error otherwise|
|`.insert(<index>, <thing to insert>)`|Inserts the given thing at the wanted numerical index.|
|`.pop(<index>)`|Removed the item at the given index from the list and returns it.|
|`.remove(<thing to nuke>)`|Removes the first matching item from the list. Chucks an error if it doesn't exist.|
|`.sort(reverse=<True or False>, key=<function to sort by>)`|Sorts the given list in place. Uses the function passed in as the `key` argument to choose how to <br> <br> sort - if not passed in, defaults to alphabetical. If `reverse=True` is passed in, sorts descending; <br> <br> default is ascending. <br> <br> <br> NOTE: it's easiest to pass in the arguments here using the parameter names, not positionally!|

Please note though - using `.index(<item in list>)` to check if an item is in a list isn't a good way of going about it! Instead, we can use the `in` operator to check if a certain element belongs to the list!

````python
# Demonstration of Some List Methods

list_of_nums = [-3, 383, 3.1415, 789, -1110]

# List methods modify the original list!!
list_of_nums.append(777)
print(list_of_nums)

print(f"Index of -3: {list_of_nums.index(-3)}")

list_of_nums.remove(-3)
print(list_of_nums)

# .sort() has a default value for the sort type parameter of sorting
# in ascending numerical order for numbers, and alphabetical order for
# strings.
list_of_nums.sort()
print(list_of_nums)

if 21 in list_of_nums:
  print("21 is in the list!")
else:
  print ("21 is not in the list!")
````

I do want to point out a Python quirk here. You know how at the start of this part I made a big deal of that `num_1 = num_2` wouldn't irrevocable link `num_1` and `num_2` together? The inverse is true with more complicated data types, like lists, dictionaries, and similar things. If I want to copy a list, just doing `list_2 = list_1` isn't sufficient - it just treats `list_2` as another name for the same `list_1`. You NEED to use the `.copy()` method!! See an example in code below!

````python
# Lists - How To Copy
list_1 = ["a", "b", "c", "d"]
list_2 = list_1
list_3 = list_1.copy()

list_1.remove("b")

# list_2 and list_1 both refer to the same list under the hood,
# not two copies - changing list_1 changes list_2!!
print(f"list_2: {list_2}")

# list_3 is an independent copy of list_1 - modifying list_1 does
# not affect it!!
print(f"list_3: {list_3}")
````

We can also put more complicated data types inside lists - for example, lists inside lists! In this case, indexing in is done as shown below!

 ## Loops Part 2 - For Loops

Let's say I wanted to iterate through a list. One way that we could go about it is using a while loop to iterate through 0 to the length of the list, and use that index to index into the list.

To get the number of entries in a list, we can use Python's inbuilt `len()` function! This function is also useful for getting the length of a string!

````python
# A perfectly working, though slightly clunky, way of iterating through a list!
list_of_nums = [-3, 383, 3.1415, 789, -1110]

index = 0
while index < len(list_of_nums):
  print(list_of_nums[index])
  index += 1
````

However, in Python there exists another type of loop that automatically iterates through the members of a given list! It's called a for loop, and for things like just iterating through a list, it's a less clunky option or going about things!

````python
list_of_nums = [-3, 383, 3.1415, 789, -1110]

# For loops iterate through every element of a list sequentially!!
# Note how much less setup there is to get the same result.
for num in list_of_nums:
  print(num)
````

Generally, if you don't care about the index itself, then using a for loop is a better choice than a while loop for iterating through a list. If for some reason you do need to work with the index (e.g you want to compare two lists manually, element by element), then a while loop would be better; however, this is a rarer case.

We can also use for loops to traverse sequences of numbers like how we use while loops! The `range()` function generates a list from the given start value to just before the stop value, with an optionally provided step size.

````python
# Step size is optional, has a default value of 1.
# Starts at 1, keeps going while less than 4.
for index in range(1, 4):
  print(index)

print("====")

# Starts at 1, keeps going while less than 6, increments
# by 2 each time.
for index in range(1, 6, 2):
  print(index)
````

## Dictionaries

Ok, we have lists, but what if I want to bundle related info together by category? For example, if I am working with a student's zID, name, phone number, whether they've passed a certain course or not - I can store them all as unrelated variables, sure, but having some way of grouping them together would be helpful.

Dictionaries let us do that! The syntax to set them up is in the form `<variable name> = {<key 1 name>=<value 1>, <key 2 name> = <value 2>, ..., <key n name>=<value n>}`. Please note, you **MUST USE CURLY BRACKETS**, else Python won't treat it as a dictionary! Key names are usually text strings (though you can use things like integers as keys in a dictionary, it's just far less common!). Values can be anything - simple types like text or numbers, or complex ones like lists or other dicts.

An example of a sample dictionary is shown below, and how we can access elements in it!

````python
# Sample Dictionary Set Up

student_info_dict = {
    "first_name": "Gatorman",
    "last_name": "McDuff",
    "zID": 1234567,           # dictionary can hold different data types!!
    "chem_student": True,
    "completed_courses": ["CHEM1031", "CHEM1041", "CHEM2041", "PHYS1121"]
}

# We access the contents of a dictionary using square bracket notation as well,
# passing in the key we want inside the square brackets, and getting that key's
# value as the printed output!
print(student_info_dict["first_name"])

# We can store more complex things in dicts too - like lists in lists. Accessing
# it evaluates left to right!
print(f"First completed course: {student_info_dict["completed_courses"][0]}.")
````

Like with lists, dictionaries have a lot of methods we can use to manipulate them! Not all update the original dictionary though, and whether a given method updates or not is specified below.

|Method Name + Syntax|Description|
|-|-|
|`.pop(<key name>)`|Removes the key-value pair that matches the given key from the ditionary|
|`.update(<dict to merge with original>)`|Adds the keys and values of the passed in dictionary to the original!|
|`.copy()`|Returns a copy of the dictionary (like lists, just doing `dict_1 = dict_2` doesn't work!)|
|`.keys()`|Returns *almost* a list of the keys in a dict. Can be iterated through with a for loop just like a list! <br> <br> If you want to use list methods on this though, need to convert to list first by passing the return <br> <br> into the `list()` function.|
|`.values()`|Returns *almost* a list of the values in a dict. Can be iterated through with a for loop just like a list!|
|`.items()`|Returns *almost* a list of the keys and values in the format `[(<key 1>, <value 1>), (<key 2>, <value 2>), ...]`. <br> <br> Can be iterated through with a for loop like a list, but you need to unpack like you do for a function with multiple returns! <br> <br> (see the example below)|

If you want to check if a certain key or value is in the dictionary, the easiest way is using the `in` operator on either `.keys()` or `.values()`. If you want to get the number of items in a dictionary, use the inbuilt `len()` function in Python like you would with lists!

````python
info_dict = {
    "name": "Natasha Brown",
    "phone_num": "+61 789 123"
}

# Don't forget the curly braces!
# info_dict will be updated with the key-value pairs we pass in below.
info_dict.update({
    "car_colour": "blue",
    "car_roadworthy": False,
})

print(info_dict)

# Removes the key "phone_num" and associated value from the dict!
info_dict.pop("phone_num")
print(f"{info_dict}\n")

# Remember - must unpack!
for key, value in info_dict.items():
  print(f"Key: {key}, Value: {value}")

print()

for key in info_dict.keys():
  print(f"Key: {key}")

print()

if "Natasha Brown" in info_dict.values():
  print("Natasha is in the dict as a value!")
````

You know how I said that `.keys()`, `.items()`, and .`values()` return things which are almost like lists, but not quite? Basically, they return what is known as a "view object". All it means is if I run `values = some_dict.values()`, and then add another key-value pair, or update the value of a certain key, `values` automatically updates! If it returned a normal list it wouldn't be able to do that.

However, if you want to make any modifications (e.g appending things or removing things in those lists for whatever reason), you need to convert to a list first! Please note, modifications to those lists won't affect the original dictionary.

````python
info_dict = {
    "name": "Natasha Brown",
    "phone_num": "+61 789 123"
}

# If we want to append something to the values list for whatever reason, we need
# to convert to a list first!!

values_list = list(info_dict.values())
values_list.append("blah")
print(values_list)
````

````python
info_dict = {
    "name": "Natasha Brown",
    "phone_num": "+61 789 123"
}

# If we don't do this, it will error - the "view objects" just lack a lot of
# list-asociated methods.
values_list_2 = info_dict.values()
values_list_2.append("blah")
````

## Brief Interlude - Let's Talk About Constants

A lot of the time when we program, we have random numbers or values that are important - for example, maybe I need to use a known scientific constant in a calculation, or I need to define some custom range of wavelengths to work with. There's both common constants that are defined by themselves, and a database of other constants accessible via the following provided methods:

|Method Name + Syntax|Description|
|-|-|
|`.value(<const name>)`|Returns the value of the constant with that name|
|`.unit(<const name>)`|Returns the units of the constant with that name|

There's a few other methods used to access things like the precision of the constants detailed in the list of available constants, however you aren't likely to need to use them.

### Custom Constants

What about if you want to define your own constants? Well, the way this is usually done in Python is using global variables, specifically written in all caps with underscores (called "screaming snake case"), that we define at the very start of our code, right under the imports.

````python
# Example of Python Constants
NUM_SCANS_PER_MINUTE = 8    # all caps indicates we treat this as a constant value!

scan_num = 4                # not all caps - when we code, we can reassign it.
````

If we want to use a constant in a function, we can just pass it in, because if we're coding correctly, we don't run into the issue of the constant having different values at different execution times and changing the function's behaviour. Since the constant should never be reassigned to another value in our code and only used to assign other things or compare things against, it should be fine!

````python
# Examples of Allowed and Not Constants
NUM_ELECTRONS_IN_CARBON = 6

num_elec = NUM_ELECTRONS_IN_CARBON    # allowed - not modifying the value of the const!
num_2 = 5

if num_2 < NUM_ELECTRONS_IN_CARBON:   # allowed - not modifying the value of the const!
  print("num 2 has fewer electrons than carbon!")

NUM_ELECTRONS_IN_CARBON = 3           # Python lets this happen, but we SHOULDN'T DO THIS - it isn't a constant
                                      # throughout the execution of our code anymore!!!
````

The problem is, Python doesn't look at the constant, see that it's in all caps, and go "oh we can never modify it". It just treats it as any other variable, and it WILL let you modify it after you first set it. It's really easy to write something like `SOME_CONST = some_var` instead of `some_var = SOME_CONST` and then have your code catch fire and you have no idea why! Python doesn't natively let us create constants that we can't modify after we first set them up.

If you trust yourself to not accidentally write a bug like that, this method works perfectly fine, and is the way things are usually handled. If you want to guarantee that a constant you make will actually be immutable (aka impossible to change after assignment), though, you can use the `setconstant` package for that! (There are ways to set up immutable constants in Python, but not without a ton of work, and this is an easier solution.)

Here's a summary of the available methods and a demo of how to use them! Note - I would still recommend giving the constants names in all caps for style reasons.

|Method + Syntax|Description|
|-|-|
|`.const_i(<const name>, <const value>)`|Adds an integer constant with the provided name. Must be an integer (whole number - no dp!) or will error!|
|`.const_f(<const name>, <const value>)`|Adds a decimal (floating point) constant with the provided name. <br> <br> <br> Note - will error with something like `.const_f("MASS", 4)` because 4 doesn't have a decimal point. <br> <br> Will work with `.const_f("MASS", 4.0)` instead!|
|`.const_s(<const name>, <const value>)`|Adds a string constant with the provided name.|
|`.const_b(<const name>, <const value>)`|Adds a boolean (True/False, written exactly that way) constant with the provided name.|
|`.get(<const name>)`|Returns the value of the const if it exists, or errors|
|`.get_type(<const name>)`|Returns what type the const was given when setup|
|`is_defined(<const name>)`| Returns True/False if the given const exists in the list of consts defined like this|
|`.summary()`|Prints a summary table of all added constants.|

````python
# Using the setconstant Package

# Installing + importing in first
!pip install setconstant
import setconstant

setconstant.const_i("NUM_SCANS_PER_MIN", 8)
setconstant.const_s("WAVENUM_COL_NAME", "wavenum")

scan_num = setconstant.get('NUM_SCANS_PER_MIN')
num_2 = 5

if num_2 == scan_num:
  print("num_2 and scan_num are the same!")
else:
  print("num_2 and scan_num are not the same!")

print(f"A constant named 'Ptolemy' exists? {setconstant.is_defined("Ptolemy")}")

setconstant.summary()
````
