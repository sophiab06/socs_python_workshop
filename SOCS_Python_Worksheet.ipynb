## Variables and Basic Data Types


Variables are the backbone of any of your programs. Like in math, they're just values we give names to, values that can hold anything we want. A proper variable name in Python can only contain letters, numbers and underscores, and must start with either a letter or an underscore. Generally, in Python it's good style to keep the letters in your variable names all lowercase. Also, make sure to not name them the same as another variable you've made already, or any names used by Python or any code you've imported, else your code can run into cryptic bugs. So, for example:


The syntax for assigning and reassigning the value is of the form `variable_name = value`. Some examples of variable assignment are presented below!

Let's address a common source of confusion in coding really quickly. Consider the below bit of code. Can you tell me what the value of `num_1` is going to
be without running it?

````python
# Variable Behaviour - Can You Predict the Outputs?

num_1 = 5
num_2 = 10
num_1 = num_2
num_2 = num_2 + 5

print(f"The value of num_1 is {num_1}, and num_2 is {num_2}")
````

Rather, it just means that at that line, we set the value of `num_1` to be whatever the value of `num_2` is. Further modifications of `num_2` will not affect whatever was in `num_1`!!

Certain operations can have multiple different meanings, depending on the data type. For example, the + symbol performs regular addition on numbers of any kind, whereas for strings it joins them together. For example, run the below code and observe the output.

````python
# Sometimes Different Types Use The Same Operator in Different Ways

# + with ints or floats adds
# them together like we expect
variable_1 = 5
variable_2 = 3.1

print("Decimal '+' Result: ")
print(variable_1 + variable_2)
print()

# + with text joins it together
# (formally: concatenates it)!
variable_1 = "hello"
variable_2 = "there"

print("String '+' Result: ")
print(variable_1 + variable_2)
````

## If Statements

Currently, we don't have a way of taking multiple paths through our code, choosing to do a certain action or not, or anything similar. For example, what if I wanted to write something that took in a number, and only printed it back out if it was even? The way we do this is with what is known as "if statements", and a basic one is presented below. Try it out!

````python
# If Statement Demonstration
# Program prints whether input num is odd or even.

num = int(input("Please enter an integer: "))

if num % 2 == 0:
  print("Even")
else:
  print("Odd")

print("All done!")
````

How exactly does this work? Let's break this down line by line.

First off, the line `if num % 2 == 0:`. `num % 2` is calculated by the computer, and then is compared against the value of 0 to see if they are equal or not. Python uses `==` for comparing whether two values are equal, since using one equal sign is already reserved for assigning.

If the condition is true, we go into the first branch of the if statement, print "Even", then leave (there's nothing else to do in this branch) and go to the next shared line of code - printing "All done". If the condition fails, the code goes into the `else` branch, prints "Odd", leaves as there's nothing else to do, and once again goes to the next shared code line.



How can we tell that `print("Even")` belongs to the `if` branch, and `print("Odd")` belongs to the `else` branch? Well, Python uses the indentation of code lines to determine what belongs where.

`print("Even")` is indented in by a tab, so it belongs to the `if` branch, and only runs if we go into that branch. `else` is not indented, so doesn't belong to the `if` branch - as there's no more indented code in this section, that's the end of the `if` branch.

Likewise, `print("Odd")` is indented in by a tab, so it belongs to the `else` branch, and only runs if we go in it. As `print("All done!")` is not indented, it belongs to neither branch, so gets run always.

We can check any number of options at once in an if statement! If we want to check for > 2 branching paths, we need to use an `elif` (a contraction of "else if") statement and add an extra condition to check. We can have any number of `elif` statements. Also to note, `else` statements are completely optional; if there's no action you take if a condition isn't true, there's no need to write one. You can also have an army of `elif`s and no `else`, that is also perfectly valid!

You don't just need to check equality, too: you can check if something is greater than/less than, greater than equal to/less than equal to, and not equal as well. Some examples of all of this are presented below.

````python
# Elif + Other Testing Option Example

num = float(input("Please enter a number: "))

if num >= 10:                               # less than or equal to would be <=
  print("Bigger than or equal to 10")
elif num < 5:
  print("Smaller than 5")
elif num != 6.66:                           # if num is not equal to 6.66
  print("Somewhat small, but not demonic.")
else:
  print("😈")
````

We can also test 2 or more things in one condition by using the keywords `and`, `or` and `not`, to specify exactly what conditions we want. One thing to note: if you use several together, it is best to use parentheses to indicate exactly what logic you want. It's very easy to make mistakes otherwise, or interpret a logical condition differently to how Python evaluates it while the program is running.

“I want a bouquet with (blue cornflowers and black roses) or (yellow roses)”
is different to
“I want a bouquet with blue cornflowers and (black roses or yellow roses)”!!

````python
# Example of How to Use "or"! "and" and "not" are Used Very Similarly.
num = float(input("Please enter a number: "))

if num > 10 or num < 0:
  print("Needs more than one digit to print!")
````

## Loops Part 1 - While Loops

At the end of the loop the running program jumps back up to retest the condition, staying in this cycle until the condition finally fails.

## Functions

Often when we code, there's going to be a specific task that we need to repeat multiple times - for example, performing the same type of calculation over and over again. We could just copy and paste the same lines of code; however, this can get messy + hard to keep track of. You can forget to modify one out of 20 of those sections when updating your code, for example, and now everything's on fire and you don't know why.

A better option is to create a function - basically, a callable chunk of code you give a name to and can run whenever you want. `print()` and `input()` are some functions which we have already seen, ones that have been provided to us. An example of a user-written function is presented below.

````python
# Example of Function Implementation and Function Calling

# NOTE: these don't have to be in the same code block in Colab, BUT you do
# need to run the block containing the function before you run a block calling
# it!!



# Function Implementation

# keyword "def" used to indicate that it's a function
# Need to indent to show what lines belong to the functions and what don't!

def find_factorial(number):   # Inside the brackets you choose + specify
  index = 1                   # "parameters" - variables you pass in while
  curr_fact = 1               # calling!

  while index <= number:
    # Calculates curr_fact as 1 * 2 * ... * number
    curr_fact = curr_fact * index
    index = index + 1

  return curr_fact            # "return" keyword is necessary to specify what
                              # the function spits back out!




# Function Calling
num = int(input("Enter a number: "))

# Here, num is called an argument - it's what's being passed in to fill a
# parameter!
factorial = find_factorial(num)
print(f"The factorial of {num} is {factorial}.")      # we can also pass in raw
print(f"The factorial of 4 is {find_factorial(4)}.")  # data, not just variables!
````

The code inside a function doesn't run unless you specifically call it. Like with everything else we've seen so far in Python, we need to indent to show what lines of code belong to a function, and which lie outside of it. When writing a function you can give it any name that isn't currently in use by another function or variable (i.e, don't write another function named `print()`).

(Something to note - parameters are the variables we specify when we define the function, but arguments are what we pass in. This can be easy to confuse! In the above, `number` is a parameter, whereas `num` and `4` are arguments because they're the things being passed into the function. ### Functions With Multiple Parameters


When we create a function, we can give it any number of parameters (variables we pass in when we call the function). However, if we have more than one variable, there are multiple ways of going about actually passing in the values we want the parameters to have when we call.

The easiest is by position: for example, calling a function with a header of
```python
def raise_to_power(number, exponent):
```
as
```python
result = raise_to_power(5, 2)
```
sets the value of number to 5 and the value of exponent to 2. (Here, `number` and `exponent` are function parameters, and `5` and `2` are arguments to the function!)

We can also pass in the arguments by the names of the parameters; the next line is equivalent to the one we just wrote above!
```python
result = raise_to_power(number=5, exponent=2)
```

When we pass in arguments by parameter name, the order of them doesn't actually matter. If there are any arguments you are passing in by order, make sure to do that before any arguments you pass in by name! The below would return the same as the prior two function calls.
```python
result = raise_to_power(exponent=2, number=5)
```

Function parameters also can have default values. If I change the function header to the following,
```python
def raise_to_power(number, exponent=1):
```
then if I only pass in an argument for number, exponent assumes a default number of 1 and the function can still run as normal. However, if I was to change the function header to the following
```python
def raise_to_power(number=2, exponent):
```
and only pass in one argument positionally, then it would change the value of number, not exponent, and the function would crash. Make sure to put any parameters that have default values after parameters that do not when writing the function definition!

Working examples of all of the above can be seen below. Run them and see!

````python
# Multiple Parameter Demonstration

# You can just do this using **, but this is to demonstrate a function
# with multiple parameters!
def raise_to_power(number, exponent=1):
  index = 1
  curr_raised_num = number

  while index < exponent:
    curr_raised_num = curr_raised_num * number
    index = index + 1

  return curr_raised_num


# Raises 5 to the 2nd power - passing arguments in using parameter order! (Prints 25)
print(f"raise_to_power(5, 2) = {raise_to_power(5, 2)}")

# Order doesn't matter if we're passing in using the parameter name! (Prints 25)
print(f"raise_to_power(number=5, exponent=2) = {raise_to_power(number=5, exponent=2)}")
print(f"raise_to_power(exponent=2, number=5) = {raise_to_power(exponent=2, number=5)}")

# This will just print 5 since the default value for exponent is 1
print(f"raise_to_power(5) = {raise_to_power(5)}")
````

### Functions and Variable Scope

Any variables that we define inside a function cannot be accessed after we return from a function - whatever data got put in them last time we called the function gets scrubbed. If you need to access multiple things after a function's return, you need to return multiple things! You can do so simply using commas like in the example below:

````python
# Multiple Variable Return Demonstration

# A randomiser function that returns two
# variables at once!
def randomiser_function(num_1, num_2):
  output_1 = num_1 * num_2 + num_1/num_2
  output_2 = num_2 ** num_1 * (num_1 - num_2)

  return output_1, output_2

# output_1 goes into random_1, output_2
# goes into random_2
# Order MATTERS here - must match up!
random_1, random_2 = randomiser_function(4, 2)
print(f"random_1 = {random_1}")
print(f"random_2 = {random_2}")
````

You *can*, however, access variables that are defined outside of a function inside one. However, **PLEASE DON'T**. Doing this is a very easy way to write code that accidentally depends on some external variable you forgot about, that may work fine one time you run the notebook but fall apart once you reload it. If you need to access a variable from outside of a function, just add an extra parameter to the function + pass it in.

 ## Loops Part 2 - For Loops

Let's say I wanted to iterate through a list. One way that we could go about it is using a while loop to iterate through 0 to the length of the list, and use that index to index into the list.

To get the number of entries in a list, we can use Python's inbuilt `len()` function! This function is also useful for getting the length of a string!

````python
# A perfectly working, though slightly clunky, way of iterating through a list!
list_of_nums = [-3, 383, 3.1415, 789, -1110]

index = 0
while index < len(list_of_nums):
  print(list_of_nums[index])
  index += 1
````

However, in Python there exists another type of loop that automatically iterates through the members of a given list! It's called a for loop, and for things like just iterating through a list, it's a less clunky option for going about things!

````python
list_of_nums = [-3, 383, 3.1415, 789, -1110]

# For loops iterate through every element of a list sequentially!!
# Note how much less setup there is to get the same result.
for num in list_of_nums:
  print(num)
````

Generally, if you don't care about the index itself, then using a for loop is a better choice than a while loop for iterating through a list. If for some reason you do need to work with the index (e.g you want to compare two lists manually, element by element), then a while loop would be better; however, this is a rarer case.

We can also use for loops to traverse sequences of numbers like how we use while loops! The `range()` function generates a list from the given start value to just before the stop value, with an optionally provided step size. Note - it can only take integers as the input!

````python
# Step size is optional, has a default value of 1.
# Starts at 1, keeps going while less than 4.
for index in range(1, 4):
  print(index)

print("====")

# Starts at 1, keeps going while less than 6, increments
# by 2 each time.
for index in range(1, 6, 2):
  print(index)
````

## Dictionaries

Ok, we have lists, but what if I want to bundle related info together by category? For example, if I am working with a student's zID, name, phone number, whether they've passed a certain course or not - I can store them all as unrelated variables, sure, but having some way of grouping them together would be helpful.

Dictionaries let us do that! The syntax to set them up is in the form `<variable name> = {<key 1 name>=<value 1>, <key 2 name> = <value 2>, ..., <key n name>=<value n>}`. Please note, you **MUST USE CURLY BRACKETS**, else Python won't treat it as a dictionary! Key names are usually text strings (though you can use things like integers as keys in a dictionary, it's just far less common!). Values can be anything - simple types like text or numbers, or complex ones like lists or other dicts.

An example of a sample dictionary is shown below, and how we can access elements in it!

````python
# Sample Dictionary Set Up

student_info_dict = {
    "first_name": "Gatorman",
    "last_name": "McDuff",
    "zID": 1234567,           # dictionary can hold different data types!!
    "chem_student": True,
    "completed_courses": ["CHEM1031", "CHEM1041", "CHEM2041", "PHYS1121"]
}

# We access the contents of a dictionary using square bracket notation as well,
# passing in the key we want inside the square brackets, and getting that key's
# value as the printed output!
print(student_info_dict["first_name"])

# We can store more complex things in dicts too - like lists in lists. Accessing
# it evaluates left to right!
print(f"First completed course: {student_info_dict["completed_courses"][0]}.")
````

Like with lists, dictionaries have a lot of methods we can use to manipulate them! Not all update the original dictionary though, and whether a given method updates or not is specified below.

|Method Name + Syntax|Description|
|-|-|
|`.pop(<key name>)`|Removes the key-value pair that matches the given key from the ditionary|
|`.update(<dict to merge with original>)`|Adds the keys and values of the passed in dictionary to the original!|
|`.copy()`|Returns a copy of the dictionary (like lists, just doing `dict_1 = dict_2` doesn't work!)|
|`.keys()`|Returns *almost* a list of the keys in a dict. Can be iterated through with a for loop just like a list! <br> <br> If you want to use list methods on this though, need to convert to list first by passing the return <br> <br> into the `list()` function.|
|`.values()`|Returns *almost* a list of the values in a dict. Can be iterated through with a for loop just like a list!|
|`.items()`|Returns *almost* a list of the keys and values in the format `[(<key 1>, <value 1>), (<key 2>, <value 2>), ...]`. <br> <br> Can be iterated through with a for loop like a list, but you need to unpack like you do for a function with multiple returns! <br> <br> (see the example below)|

If you want to check if a certain key or value is in the dictionary, the easiest way is using the `in` operator on either `.keys()` or `.values()`. If you want to get the number of items in a dictionary, use the inbuilt `len()` function in Python like you would with lists!

````python
info_dict = {
    "name": "Natasha Brown",
    "phone_num": "+61 789 123"
}

# Don't forget the curly braces!
# info_dict will be updated with the key-value pairs we pass in below.
info_dict.update({
    "car_colour": "blue",
    "car_roadworthy": False,
})

print(info_dict)

# Removes the key "phone_num" and associated value from the dict!
info_dict.pop("phone_num")
print(f"{info_dict}\n")

# Remember - must unpack!
for key, value in info_dict.items():
  print(f"Key: {key}, Value: {value}")

print()

for key in info_dict.keys():
  print(f"Key: {key}")

print()

if "Natasha Brown" in info_dict.values():
  print("Natasha is in the dict as a value!")
````

You know how I said that `.keys()`, `.items()`, and .`values()` return things which are almost like lists, but not quite? Basically, they return what is known as a "view object". All it means is if I run `values = some_dict.values()`, and then add another key-value pair, or update the value of a certain key, `values` automatically updates! If it returned a normal list it wouldn't be able to do that.

However, if you want to make any modifications (e.g appending things or removing things in those lists for whatever reason), you need to convert to a list first! Please note, modifications to those lists won't affect the original dictionary.

````python
info_dict = {
    "name": "Natasha Brown",
    "phone_num": "+61 789 123"
}

# If we want to append something to the values list for whatever reason, we need
# to convert to a list first!!

values_list = list(info_dict.values())
values_list.append("blah")
print(values_list)
````

### Custom Constants

What about if you want to define your own constants? Well, the way this is usually done in Python is using global variables, specifically written in all caps with underscores (called "screaming snake case"), that we define at the very start of our code, right under the imports.

````python
# Example of Python Constants
NUM_SCANS_PER_MINUTE = 8    # all caps indicates we treat this as a constant value!

scan_num = 4                # not all caps - when we code, we can reassign it.
````

If we want to use a constant in a function, we can just pass it in, because if we're coding correctly, we don't run into the issue of the constant having different values at different execution times and changing the function's behaviour. Since the constant should never be reassigned to another value in our code and only used to assign other things or compare things against, it should be fine!

````python
# Examples of Allowed and Not Constant Actions
NUM_ELECTRONS_IN_CARBON = 6

num_elec = NUM_ELECTRONS_IN_CARBON            # allowed - not modifying the value of the const!
atom_2_elects = 5

if atom_2_elects < NUM_ELECTRONS_IN_CARBON:   # allowed - not modifying the value of the const!
  print("Atom 2 has fewer electrons than carbon!")

NUM_ELECTRONS_IN_CARBON = 3                   # Python lets this happen, but we SHOULDN'T DO THIS - it isn't a constant
                                              # throughout the execution of our code anymore!!!
````

The problem is, Python doesn't look at the constant, see that it's in all caps, and go "oh we can never modify it". It just treats it as any other variable, and it WILL let you modify it after you first set it. It's really easy to write something like `SOME_CONST = some_var` instead of `some_var = SOME_CONST` and then have your code catch fire and you have no idea why! Python doesn't natively let us create constants that we can't modify after we first set them up.

If you trust yourself to not accidentally write a bug like that, this method works perfectly fine, and is the way things are usually handled when you code in Python. If you want to guarantee that a constant you make will actually be immutable (aka impossible to change after assignment), though, you can use the `setconstant` package for that! (There are ways to set up immutable constants in Python without importing packages, but not without a ton of work, and this is an easier solution.)

```python
high_earner_df = all_people_df.query("Income > @income_limit")

# Income is the name of a column, income_limit is the name
# of an external variable.
```

### A Note on the Almost-Lists (Series)


Series can best be described as roided up lists with a bunch of custom methods. You can index them as you would a regular Python list, and there are very close analogues for the methods lists have (e.g `.concat()` for Series is analogous to `.append()` for lists, though with slightly different arguments). Conversion from a Series to a list is easy, using the `pd.Series()` (after importing Pandas as `pd`) and `list()` functions to go in either direction.

The major benefit of series is the custom methods, especially math ones. You can use +, -, /, * and similar operations on series without needing to iterate through each element!

Some examples of useful methods series have are detailed below. All of the methods can be found [here](https://pandas.pydata.org/docs/reference/series.html).

|Method Name + Syntax|Description|
|-|-|
|`.max()`/`.min()`|Returns the maximum or minimum value in the series. <br> <br> Useful for normalising data by dividing by the maximum value, e.g for absorbance.|
|`.idxmax()`/`.idxmin()`|Returns the index of the max or min value in the series. <br> <br> Useful for e.g getting the wavenumber of a max absorbance value!|
|`.mean()`|Returns the average of the series. (`.median()`, `.mode()`, etc also exist!)|
|`.cumsum()`|Returns a new series where each entry is the cumulative sum of the entries before.|


````python
# Series Demo

nums_1 = pd.Series([1, 2, 3, 4, 5])
nums_2 = pd.Series([6, 17, 8, 9, 10])

# We can add/subtract/multiply/divide etc series together
# element by element! NOTE - only works for series of the 
# same length!
print("nums_1 * nums_2:")
print(nums_1 * nums_2)

# We can add/substract/multiply/divide etc by constant
# values as well!
print("\nnums_1 ** 2: ")
print(nums_1 ** 2)

# Example of the methods!
print(f"\nMax of nums_1: {nums_1.max()}")
print(f"\nCumulative sum of elements in nums_2: \n{nums_2.cumsum()}")
print(f"Sum of elements in nums_2: {nums_2.sum()}")

print(f"\nValue of nums_1 at the same index as nums_2's maximum: {nums_1[nums_2.idxmax()]}")
````

If you are working with things like wavenumber and absorbance data, it can be easier to just work with the columns as individual series, and use series-based methods, than modify the whole dataframe and use dataframe-based things.

One issue that we do run into when doing operations on two series is index misalignment. When operations are performed on series, it doesn't compare the first element of each series against each other, then the second, then the third. It matches elements based on the index.

````python
# Demo of Pandas doing Operations Based on Index

strings_1 = pd.Series(["a", "b", "c", "d", "e"])
strings_2 = pd.Series(["1", "2", "3", "4", "5"], index=[1, 3, 4, 5, 2])

# If the string joining was performed without caring for the
# index, we'd expect an output that'd be ["a1", "b2", "c3", "d4", "e5"]
# as a series!

# Note the NaNs when one series has an entry with an index and
# another doesn't!
print(strings_1 + strings_2)
````

This issue usually comes up after we trim data in particular. For example, say I have a reference spectrum and an experimental spectrum that started and ended at different wavenumbers. I can trim them to have the same start and end wavenumber bounds, but now if I try to do something like subtract the absorbances of the two spectra from each other, I can run into significant issues due to the indices being misaligned and Pandas not handling that properly.

To fix this, we need to update the index! We can either reindex the individual series or reindex the dataframes as a whole.

To reindex a series, we can set the `.index` attribute to a custom list. To reindex a dataframe, we can reuse the `.set_index()` method, passing in a list to act as the new index instead of a column name like before.

If you want to reset the indices to be a list of integers from 0 to the length of the dataframe or series, you can generate a list of numbers using the `range()` function!

Since series have custom indices, we can't use negative numbers for positional indexing! If you want to use positional indexing, you need to use the `.take()` method, which allows for list-style indexing by position and using negative numbers to index from the back!

````python
# Reindexing a series
# Wanted result - ["f1", "g2", "h3", "i4", "j5"]

# Let's say this was the data we loaded in for some reason!
strings_1 = pd.Series(["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"])
strings_2 = pd.Series(["1", "2", "3", "4", "5"])

# We can trim series using syntax similar to dataframes, but there is no
# query method!
strings_1_trimmed = strings_1[strings_1 >= "f"]

print("strings_1 Trimmed:")
print("Note the index numbers starting from 5, not 0!")
print(strings_1_trimmed)

print("\nstrings_1_trimmed + strings_2: Result if we don't reindex:")
print(strings_1_trimmed + strings_2)

# Reindexing to realign the strings_2 and strings_1 series for
# the final wanted result!
strings_1_trimmed.index = range(0, len(strings_1_trimmed))
strings_2.index = range(0, len(strings_2))

print("\nstrings_1_trimmed + strings_2: Result after reindexing:")
print(strings_1_trimmed + strings_2)
````

#### Point Labels and Data Legends

The recommended way to set up the legend correctly is to set labels when you plot data, and then just call the `legend()` function with no parameters before calling `show()`. You can also pass `legend()` an array of labels matching the order of plotted data, but PyPlot's documentation discourages this as if you change the order of plotting down the road, it's incredibly easy to get the labels aligned to the wrong data.

`legend()` has two more parameters worth being aware of. `loc` takes in a string specifying the location: either `"best"` to let PyPlot decide, or something like `"upper right"`, `"lower center"`, `"center left"` or `"center"` to specify a section of the graph. `fontsize` takes one of `"xx-small"`, `"x-small"`, `"small"`, `"medium"`, `"large"`, `"x-large"`, or `"xx-large"`, and is useful for rescaling large legends easily.

Also, putting things like the axis limits in constants can be a good idea!

````python
# Legend Demonstration + Function For Plotting Finalisation Demo.

# CONSTANTS - usually should be located after all of the imports!
# Here as an example
X_LIM_LOWER = 200
X_LIM_HIGHER = 600
Y_LIM_LOWER = 0
Y_LIM_HIGHER = 1.04

X_MINOR_TICK_STEP = 10
Y_MINOR_TICK_STEP = 0.04

# Finalises the plotting (title, axis labels, axis limits, axis ticks)
# so we don't need to have this copied everywhere!
def plot_finalise_and_show(title_extra_string):
  plt.title(fr"$\beta$-Carotene UV-Vis Absorbance {title_extra_string}")
  plt.xlabel("Wavelength / nm")
  plt.ylabel("Absorbance / a.u.")

  # It is often a good idea to put these in constants to make it
  # easier to modify in one place at the top of the file if need be,
  # especially with large files!
  plt.xlim(X_LIM_LOWER, X_LIM_HIGHER)
  plt.ylim(Y_LIM_LOWER, Y_LIM_HIGHER)

  plt.xticks(np.arange(X_LIM_LOWER, X_LIM_HIGHER, X_MINOR_TICK_STEP), minor=True)
  plt.yticks(np.arange(Y_LIM_LOWER, Y_LIM_HIGHER, Y_MINOR_TICK_STEP), minor=True)

  plt.show()

  return

# Recommended method - pass labels into plot!
# NOTE - we can use formatted strings even here!
plt.plot(carot_wavelength, carot_absorbance_norm, label=r"$\beta$-Carotene", color="orange")
plt.legend()

plot_finalise_and_show("(With Recommended Legend Setup)")

# Discouraged method - pass labels into legend()
plt.plot(carot_wavelength, carot_absorbance_norm, color="OrangeRed")
plt.legend([r"$\beta$-Carotene",])

plot_finalise_and_show("(With Discouraged Legend Setup)")

# Setting location and font size
# Recommended method - pass labels into plot!
plt.plot(carot_wavelength, carot_absorbance_norm, label=r"$\beta$-Carotene", color="darkred")
plt.legend(loc="upper left", fontsize="xx-large")

plot_finalise_and_show("(With Custom Legend Location + Size)")
````

#### Curve Fitting + Dealing with Data Uncertainty

A lot of the time we want to fit a regression curve to our data, to calculate things like kinetic rate constants. We can do this easily in Python! Let's first load in some example data, coming from some sort of noisy signal.


````python
# Regression Demo - Loading In Test Data

# Reimporting to allow for independent running of subsection!
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Constants
SIN_X_LIM_LOWER = 0
SIN_X_LIM_UPPER = 10
SIN_Y_LIM_LOWER = -3
SIN_Y_LIM_UPPER = 3.5

SIN_X_MINOR_TICK_STEP = 0.25
SIN_Y_MINOR_TICK_STEP = 0.25

noisy_sine_df = pd.read_csv("noisy_sine_data.csv")
signal_time = noisy_sine_df['Time (s)']
signal_value = noisy_sine_df['Signal (a.u.)']


# General function to format the sine graph how we want to - setting
# this up because I'll be making several versions of this graph in the demo.
# If you're only making one, you don't need to bother!
def sinusoidal_plot_finalise(legend_size="medium"):
  plt.legend(fontsize=legend_size)

  plt.title("Signal vs Time")
  plt.xlabel("Time / s")
  plt.ylabel("Signal / a.u.")

  plt.xlim(SIN_X_LIM_LOWER, SIN_X_LIM_UPPER)
  plt.ylim(SIN_Y_LIM_LOWER, SIN_Y_LIM_UPPER)

  # Want to have minor ticks on non-int values - need to use arange!
  plt.xticks(np.arange(SIN_X_LIM_LOWER, SIN_X_LIM_UPPER, SIN_X_MINOR_TICK_STEP), minor=True)
  plt.yticks(np.arange(SIN_Y_LIM_LOWER, SIN_Y_LIM_UPPER, SIN_Y_MINOR_TICK_STEP), minor=True)

  return

# Handles the formatting and the showing together!
# Separated from finalisation function for saving demo.
def sinusoidal_plot_fin_and_show(legend_size="medium"):
  sinusoidal_plot_finalise(legend_size)
  plt.show()


plt.plot(signal_time, signal_value, label="Raw Data", marker="+", linestyle="none", color="#2EBFA5")
sinusoidal_plot_fin_and_show()
````

Looking at the data, it looks very sinusoidal. Let's make a sinusoidal regression!

First, we need to create a fitting function that matches the type of curve we're expecting. The order of parameters here matters: the first parameter should be where we pass our x-axis data, and all other parameters should only be used for fitting. We can't make a function which takes in parameters that need to be held constant, or other things such as another series, without difficulty, though an easy-ish way to do this will be presented in Part 2 of the worksheet!

For example, if I'm expecting a linear relationship between the x data and the y data, I'd write a fitting function like the below:
```python
def linear_fit(input_series, m, b):
  return m * input_series + b
```

An exponential fit could be written like this:
```python
import numpy as np

def exponential_fit(input_series, a, b, c):
  return a * np.exp(input_series * b) + c
```

These functions don't need to single line ones, I'm writing it like so for brevity!

Since we are expecting a sinusoidal fit here, we can write a sinusoidal fitting function as follows (don't forget to run this cell!):


````python
# Regression Demo - Sinusoidal Fit Func

# Applies a sinusoidal transformation on the input data series, per
# the given parameters. Returns an output series!
def sinusoidal_fit(input_series, a, b, c, d):
  return a * np.sin(input_series * b + c) + d

print("You've ran this cell, and Python now knows the sinusoidal_fit() function"
  " exists!")
````

Once we've done this, we can use the `curve_fit()` function from `scipy.optimize` to automatically find the best values for the parameters, e.g `a`, `b`, `c` and `d` above!

`curve_fit()` takes 3 arguments positionally. The first one is the name of the fitting function, the second the x-axis data, and the third our experimental y-axis data. There's also a few parameters we specify by keyword that are important:
- `p0`: Takes in a list with our best guess for the parameters, in the same order as the parameters. For example, for a fitting function with a header of `def some_fit(input_series, a, b, c):`, and `p0=[1, 2, 3]` passed into `curve_fit`, the initial guess for `a` is 1, `b` is 2 and `c` is 3. We should always supply this! It's best to give accurate-ish estimates for the parameters, else optimisation can fail or optimise wrong. At the minimum, make sure that you're passing in starting parameter guesses of the correct sign (positive/negative)!
- `sigma`: Takes in a series or list that contains the standard deviations of the y data in order. We usually don't have this data, but if it is available it's best to pass it in to get a more accurate fit!

`curve_fit()` returns two values. In order, they are:
- `popt`: a series-like object (we can just treat as a series), containing the optimal values for the parameters in the same order as we passed in in `p0`.
- `pcov`: a 2D matrix known as a covariance matrix, containing info about how the parameters vary in relation to each other (e.g a higher parameter `a` value is associated with a higher parameter `b` value).

Let's optimise the parameters for our sinusoidal function, and then we can talk about how to use `popt` and `pcov`!


````python
# Regression Demo - Parameter optimisation
from scipy.optimize import curve_fit

# Getting the optimal parameters + covar matrix for a sinusoidal
# fit of the input data!
# We have no info regarding the standard deviations, so we won't pass anything
# in for that.
signal_popt, signal_pcov = curve_fit(sinusoidal_fit, signal_time, signal_value, p0=[2, 1, 1, 1])

print("You've ran this cell, and Python now knows of the signal_popt and "
  "signal_pcov variables!")
````

To see what the optimal value for a parameter is, we can now just index into `popt`! We can also now pass those optimal values in along with our x-axis data into our fitting function to obtain theoretical y-axis values, and then plot that to see our regression curve.

We can pass in the parameters one by one by passing `popt[0], popt[1], popt[2]...` as arguments, however an easier option is by using Python's splat operator! Putting a * in front of a list when passing the list into a function makes python pass each of the elements of the list into the function as individual arguments in order, instead of passing the list as one argument.

Note - if you are working with a non-linear regression and only a handful of x-axis data points, just passing in the x-axis data to generate the regression y data would give you a very choppy curve. Instead, use `range()` or `numpy.arange()` to generate a list of many, equally spaced x-values, and pass those in with the optimal parameters to generate many theoretical y-values instead, that will look smoother when plotting!

````python
# Regression Demo - Plotting Theoretical Y-axis Values

# Accessing an optimal parameter - remember, order in the sinusoidal_fit
# function is a, b, c, d, so signal_popt[0] matches a and so on!
print(f"The optimal value for parameter d is {signal_popt[3]}.")

# Getting theoretical y-axis data + using the splat (*) parameter!
# Identical to calling sinusoidal_fit(signal_time, signal_popt[0], signal_popt[1], signal_popt[2], signal_popt[3]),
# and works regardless of popt list's length.
theoretical_signal_value = sinusoidal_fit(signal_time, *signal_popt)

# Remember to format the numerical values to a low number of decimal places
# so the legend doesn't become comically long due to 10dp numbers!
# We can use r along with f here to add math formatting if needed!
regr_label = f"Regression: y = {signal_popt[0]:.2f}sin({signal_popt[1]:.2f}x + {signal_popt[2]:.2f}) + {signal_popt[3]:.2f}"

# Plotting the raw data and the regression curve!
plt.plot(signal_time, signal_value, label="Raw Data", marker="+", linestyle="none", color="#2EBFA5")
plt.plot(signal_time, theoretical_signal_value, label=regr_label, color="#216869")

sinusoidal_plot_fin_and_show("x-small")
````

The only part of `pcov` that we really care about is the diagonal, as it contains the variances for each of the parameters, in the same order as in `popt`. For example, the variance of `popt[3]` is `pcov[3][3]`. To isolate just this diagonal in particular, we can run `numpy`'s `diagonal(<some matrix>).copy()` function, passing in `pcov`! This extracts a view-only view of the matrix's diagonal as a series-like structure, and then makes a copy so we can modify it further. We can then get the standard deviations of all of the parameters at once by applying a square root to the series of variances.

````python
# Regression Demo - Getting Standard Deviations

# Getting all standard deviations at once from the pcov matrix's diagonal!
signal_param_variances = np.diagonal(signal_pcov).copy()
signal_param_sds = signal_param_variances ** 0.5

print(f"The exact optimised value for parameter b is {signal_popt[1]} +/- {signal_param_sds[1]}.")

# Accessing the variance via direct indexing into the pcov matrix
print(f"The exact optimised value for parameter c is {signal_popt[2]} +/- {signal_pcov[2][2] ** 0.5}.")
````

To enter debugging mode, the easiest way is to call the function `set_trace()` from `ipdb` at the spot where we want to enter debugging mode. This is called setting a breakpoint in the code. Once we are in the debugging mode, we can step through our code line by line and do a bunch of other helpful actions!

`set_trace()` takes two optional arguments by keyword:
  - `context`: Takes in an integer, specifying how many lines of code are showed (including the current line in the middle) when we go into debug mode. Default is 3 (one above and below).
  - `cond`: Takes a True/False (boolean) value, usually passed in the form of a comparison (e.g `cond=(name == "Francois")`, or `cond=(num_2 > 7)`). Once we're in debugging mode, there is a bunch of commands we can run to get a better understanding of the current state of our code. A summary table is below!

````python
# ipdb Demo - Working Code
# Play around with the debuggers by uncommenting whichever you want,
# and see what the commands do!
# This code is just some fun math sequences that we can play around with!

# e isn't a scipy const - need to use numpy for that!
import numpy as np

# Calculated num!, returning it!
def factorial(num):
  curr = 1
  index = 1

  while index <= num:
    curr = curr * index
    index = index + 1

  return curr

# Finds the nth Catalan number (https://en.wikipedia.org/wiki/Catalan_number)
# and returns it.
def catalan_num(num):
  # # Debugger to play around with!
  # # Note how it outputs 2 lines in front and back, not 1!
  # ipdb.set_trace(context=5)

  numerator = factorial(2 * num)
  denominator = factorial(num + 1) * factorial(num)
  return int(numerator / denominator)

# Finds the derangement of a set of size num (aka the number of ways that
# num items can be rearranged with nobody in the same spot as they were
# before) and returns it! (https://en.wikipedia.org/wiki/Derangement)
def derangement_num(num):
  # # Debugger to trace!
  # # Note how it only fires if the cond is matched!
  # ipdb.set_trace(cond=(num > 4))

  # Special cases
  if num == 0:
    return 1
  if num == 1:
    return 0

  # General case
  curr = 0
  last = 0
  second_last = 1
  index = 2

  while index <= num:
    # Calculate curr
    curr = (index - 1) * (last + second_last)

    # Update last + 2nd last
    second_last = last
    last = curr
    index += 1

  return curr


###############################
# CODE EXECUTION STARTS HERE! #
###############################

curr_num = int(input("Please enter a number between 0 and 10 inclusive (enter -1 to stop): "))

while curr_num != -1:
  # # Debugger here to play around with!
  # ipdb.set_trace()

  if curr_num >= 0 and curr_num <= 10:
    print(f"n! is {factorial(curr_num)}.")
    print(f"The nth Catalan number is {catalan_num(curr_num)}.")
    print(f"The nth derangement is {derangement_num(curr_num)}.")
    print(f"\nThe nth derangement at its limit is equivalent to n!/e.")
    print(f"n!/e for this number is {factorial(curr_num)/np.exp(1)}.")

    if derangement_num(curr_num) != 0:
      derangement_nfact_rat = factorial(curr_num)/(np.exp(1) * derangement_num(curr_num))
      print(f"n!/e is {100 * derangement_nfact_rat:.5f}% of the derangement for n = {curr_num}.")
    else:
      print(f"n!/e is infinitely larger than the nth derangement as the latter is 0.")

  else:
    print("Number out of bounds - please enter a new one!")

  curr_num = int(input("\nPlease enter a number between 0 and 10 inclusive (enter -1 to stop): "))
````
